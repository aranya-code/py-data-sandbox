#### Add/Remove column

```python
# Adding a new column
data['NewColumn'] = 'Aranya'
```

```python
# Dropping a column
data.drop(data['region_order'], inplace=True)
```

```python
data.head()
```

```python
# Renaming a column
data.rename(columns={'ZIP':'Zip_Code'})
```

```python
# Dropping a column
data.drop(columns='NewColumn')
```

```python
# Adding new column with string function
data['College_First_Part'] = data['INSTNM'].str.split(' ').str[0]
```

```python
data['College_First_Part']
```

```python
data.head()
```

```python
# Adding a custom column with evalution
data['Enrollment_Category']= data['undergraduate_enrollment_primary'].apply(lambda x: 'low' if x < 500 else ('medium' if x<=5000 else 'high'))
```

```python
data.head()
```

#### Aggregating data

```python
# Returing the count of uniques values of a column
data['Enrollment_Category'].value_counts()
```

```python
# Grouping by State 
data.groupby(['STABBR'])['INSTNM'].count()
```

```python
# Using aggregate groupby
data.groupby(['STABBR']).agg({'undergraduate_enrollment_primary': sum, 'region_order': sum})
```

#### Advanced Functionality

```python
# Ranking data based on a column
data['Ranked_LATITUDE'] = data['LATITUDE'].rank()
```

```python
# Ranking data based on ranked result
data.sort_values(['Ranked_LATITUDE'], ascending=False)
```

```python
# Using cumilative sum
data['cumilative_region_order'] = data['region_order'].cumsum()
```
