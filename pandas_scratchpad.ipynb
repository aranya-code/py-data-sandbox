## Loading DataFrame from Files

```python
# Reading data from CSV file
data= pd.read_csv('./DataSet/college_value_tableau_ready.csv')
```

```python
# Get top 5 rows
data.head()
```

```python
# Fetching a column from df
data.CITY
```

```python
# Fetch rows using loc(column #, row #)
data.loc[3:9]
```

```python
# Fetching a column using loc
data.loc[3:9, 'ZIP']
```

```python
# Fetching multiple columns by loc
data.loc[3:12, ['INSTNM', 'admission_rate_band']]
```

```python
data.iloc[3:12, 0:2]
```

```python
# Accessing a single column
data.CITY
```

```python
# Sorting data dataframe
data.sort_values('ZIP')
```

```python
# Looping through the dataframe
for index, row in data.iterrows():
    print(index)
    print(row)
    print('\n\n')
```

```python
# Filtering dataframe using columns
data.loc[data['admission_rate_status'] == 'Available', ['UNITID', 'CITY', 'ZIP']][0:10]
```

```python
# Filtering dataframe using contains keyword
data[data['CITY'].str.contains('Mont')]
```
