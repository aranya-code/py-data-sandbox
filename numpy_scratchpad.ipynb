#### Basics of Numpy

```python
# Importing Numpy
import numpy as np
import sys
```

```python
# Creating a 1D Array
a = np.array([1,2,3])
print(a)
```

```python
# Creating a 2D Array
b = np.array([[4,5,6],[7,8,9]])
print(b)
```

```python
# Getting the dimension
a.ndim
```

```python
# Getting the shape
b.shape
```

```python
# Data Type
a.dtype
```

```python
# Get size
b.itemsize
```

#### Accessing the element

```python
# Get a specific element [Row, Column]
print(b[1,2]) 
print(b[0,-1]) # With negative index
```

```python
# Get a specific row
b[0,:]
```

```python
# Get a specific column
b[:, 1]
```

```python
# Getting specific element by index
b[0, 0:2:1]
```

```python
# Changing an element
a[1]=5
a
```

#### 3D Array

```python
c = np.array([[[1,2],[3,4]],[[5,6],[7,8]]])
c
```

```python
# Fetching a specific element
c[1,1,0]
```

#### Initializing different types of Arrays

```python
# All 0s in matrix
np.zeros(5)
```

```python
# All 1s in martrix
np.ones((4,2,2), dtype=int)
```

```python
# Any number full method
np.full((2,3),9)
```

```python
# Any number full_like method
np.full_like(c, 5)
```

```python
# Random decimal numbers
np.random.rand(4,5)
```

```python
# Random integer values
np.random.randint(2,7, size=(3,3))
```

```python
# Identity matrix
np.identity(5)
```

```python
# Repeating an array
arr = np.array([[1,7,9]])
arr_repeat = np.repeat(arr,3, axis=0)
arr_repeat
```

#### Reorganizing arrays

```python
before = np.array([[1,2,3,4], [5,6,7,8]])
before
```

```python
after = before.reshape(4,2)
print(before)
print(after)
```

```python
# Vertically stack arrays
v1 = np.array([1,2,3,4])
v2 = np.array([5,6,7,8])

np.vstack([v1,v2])
```

```python
# Horizontally stack arrays
h1 = np.ones([2,4])
h2 = np.zeros([2,2])

np.hstack([h1,h2])
```
